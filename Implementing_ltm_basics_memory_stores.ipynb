```python
from langgraph.store.memory import InMemoryStore
```

```python
# stroing it in a variables
store = InMemoryStore()
```

# creating of the namespace and memory

```python
#creating the namespace 
namespace1 = ("user", "u1")

# adding the data in this namespace
store.put(namespace1, "1", {"data": "user likes pizza"})
store.put(namespace1, "2", {"data": "user likes dark mode in PC"})
```

```python
#creatin the another namepspace
namespace2 = ("user", "u2")

# adding the data in this namepsace
store.put(namespace2, "1", {"data": "user like python programming"})
store.put(namespace2, "2", {"data" : "user uses most of the glue code in python"})
```

# Now we will retrieve this memeory data which is created

```python
# reteriving one by one 
store.get(namespace1, "1")
```

```python
store.get(namespace1, "2")
```

```python
# for retriving all the memory in one go 
items=store.search(namespace1)
for item in items:
    print(item.value)
```

```python
items = store.search(namespace2)
for item in items:
    print(item.value)
```
